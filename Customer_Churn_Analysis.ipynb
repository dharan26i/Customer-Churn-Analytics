```python
df["churn"].value_counts()
```

```python
churn_rate = (df["churn"] == "Yes").mean() * 100
churn_rate
```

```python
total_customers = len(df)

churned_customers = (df["churn"] == "Yes").sum()

retained_customers = (df["churn"] == "No").sum()

print("Total Customers:", total_customers)
print("Churned Customers:", churned_customers)
print("Retained Customers:", retained_customers)
print("Churn Rate:", round(churn_rate, 2), "%")
```

```python
contract_analysis = (
    df.groupby("contract")
      .agg(
          total_customers=("customer_id", "count"),
          churned_customers=("churn", lambda x: (x == "Yes").sum())
      )
      .reset_index()
)

contract_analysis["churn_rate"] = (
    contract_analysis["churned_customers"]
    / contract_analysis["total_customers"]
    * 100
)

contract_analysis
```

```python
contract_analysis.sort_values("churn_rate", ascending=False)
```

```python
df["tenure_group"] = pd.cut(
    df["tenure"],
    bins=[-1, 12, 24, 48, 72],
    labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
)

df.groupby("tenure_group", observed=True)["churn"].apply(
    lambda x: (x == "Yes").mean() * 100
).round(2)
```

```python
payment_analysis = (
    df.groupby("payment_method")
      .agg(
          total_customers=("customer_id", "count"),
          churned_customers=("churn", lambda x: (x == "Yes").sum())
      )
)

payment_analysis["churn_rate"] = (
    payment_analysis["churned_customers"]
    / payment_analysis["total_customers"] * 100
).round(2)

payment_analysis.sort_values("churn_rate", ascending=False)
```

```python
internet_analysis = (
    df.groupby("internet_service")
      .agg(
          total_customers=("customer_id", "count"),
          churned_customers=("churn", lambda x: (x == "Yes").sum())
      )
)

internet_analysis["churn_rate"] = (
    internet_analysis["churned_customers"]
    / internet_analysis["total_customers"] * 100
).round(2)

internet_analysis.sort_values("churn_rate", ascending=False)
```

```python
df["monthly_charge_group"] = pd.cut(
    df["monthly_charges"],
    bins=[0, 40, 70, 100, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)

charge_analysis = (
    df.groupby("monthly_charge_group", observed=True)
      .agg(
          total_customers=("customer_id", "count"),
          churned_customers=("churn", lambda x: (x == "Yes").sum())
      )
)

charge_analysis["churn_rate"] = (
    charge_analysis["churned_customers"]
    / charge_analysis["total_customers"] * 100
).round(2)

charge_analysis
```

```python
service_columns = [
    "online_security",
    "online_backup",
    "device_protection",
    "tech_support",
    "streaming_tv",
    "streaming_movies"
]

df["total_services"] = (
    df[service_columns]
    .apply(lambda x: (x == "Yes").sum(), axis=1)
)

df[["customer_id", "total_services"]].head(10)
```

```python
avg_monthly_charge = df["monthly_charges"].mean()

avg_monthly_charge
```

```python
df["risk_segment"] = np.where(
    (df["contract"] == "Month-to-month") &
    (df["tenure"] <= 12) &
    (df["monthly_charges"] > avg_monthly_charge),
    "High Risk",
    "Other"
)

df["risk_segment"].value_counts()
```

```python
risk_analysis = (
    df.groupby("risk_segment")
      .agg(
          total_customers=("customer_id", "count"),
          churned_customers=("churn", lambda x: (x == "Yes").sum())
      )
)

risk_analysis["churn_rate"] = (
    risk_analysis["churned_customers"]
    / risk_analysis["total_customers"] * 100
).round(2)

risk_analysis
```
